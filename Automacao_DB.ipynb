```python
import os
import zipfile
import shutil

a = 1
```

```python
with zipfile.ZipFile("placas_weg.zip", 'r') as zip:
    zip.extractall("temp")

os.makedirs("placas/WEG")

arquivos_extraidos = os.listdir("temp")
arquivos_extraidos.sort()

for i in range(len(arquivos_extraidos)):
    arquivo_atual = arquivos_extraidos[i]
    novo_nome = f"Placa_WEG{a}"
    caminho_origem = f"temp/{arquivo_atual}"
    caminho_destino = f"placas/WEG/{novo_nome}.jpg"
    os.rename(caminho_origem, caminho_destino)
    a += 1

shutil.rmtree("temp")

print("Processo concluído!")
```

```python
zip_again = "placas"
new_zip = "placas"

shutil.make_archive(new_zip, 'zip', zip_again)

print(f"Pasta compactada! O arquivo {new_zip}.zip foi criado.")

from google.colab import files
files.download(f"{new_zip}.zip")
```
